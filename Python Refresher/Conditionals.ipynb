# Conditional Statement in Python and Logical Operators

##### In python, conditional statements are very important for every python project. Conditional statements are used to perform an action as long as a condition is met.
##### Conditionals in python are: "if", "elif" (else if) in other programming languages, and "else"
##### While Logical operators are used to support conditionals statement make them more robust, e.g if the user is the husband or wife, allow to get the report card. The three common logical operators in python are: "and", "or", and "not"


```python
# Application of a conditional statement that takes input from users

age = int(input("Enter your age :"))
if age >= 18:
    print("You are eligible to vote")
else:
    print("You are not allowed to vote")

# This program takes an input from the user and analyses the input to see if it meets the condition stated
```

```python
# Application of elif 

grade = float(input("Enter your grade :"))
if grade >= 70:
    print("You got an A")
elif grade >= 60:
    print("You got a B")
elif grade >= 50:
    print("You got a C")
elif grade >= 45:
    print("You got a D")
else:
    print("You got an F")

# This program takes the user's score and returns their grade from an examination
```

```python
age = int(input("Enter your age below :"))
grade = int(input("Enter your exams score: "))

if age >= 18 and grade >= 70:
    print("Welcome to the Elite class")
elif age >= 18 or grade >= 70:
    print("You got in here through either your age or score")
else:
    print("You are not eligible because your age or score didn't get you anywhere")

# Here I didn't use the not logical operator, but it acts in the opposite manner 
```
