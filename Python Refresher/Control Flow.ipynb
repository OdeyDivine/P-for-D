# Control Flows in Python

##### Loop is one of the major tool in control flow. Loops enables us execute a block of code many times as long as the condition is met
##### The are two types of loop in python which are:
- While loop
- For Loop

### While Loop
##### this type of loop will run as far as the condition is true

```python
number = 1
while number <= 10:
    print(number)
    number += 1
```

```python
# while loop with Break statement. Here we will try to guess our number
our_number = 7
while True:
    your_number = int(input("Guess our number: "))
    if your_number == our_number:
        print("Wow, you guessed right")
        break
    else:
        print("Please try again, you didn't guess correctly")

# This program continues to run as a result of the "while True" block, and then the program stops as soon as the user guesses the correct number as a result of the break statement
```

```python
# while loop with continue statement

a = 0
while a <= 20:
    a += 1
    if a % 2 == 0:
        continue
    print(a)

# this program prints only odds number from 0 to 20, note that the code snippet (if a % 2 == 0, continue) is responsible for this. When the code evaluates and sees that the modulus is 0, it means it is even; then it skips to the next
```

```python
# while loop with random. Here we will be building a game where the user try to guess the computer number which is dynamic
import random
computer_number = random.randrange(1,11)
while True:
    human_number = int(input("Guess the computer number: "))
    if human_number == computer_number:
        print("Congratulations!, you guessed the computer number")
        break
    elif human_number < computer_number:
        print(f"Too low! Try again. The correct answer was {computer_number}")
    else:
        print("Hmmm!, you didn't guess right, please try again")
```

### For Loop
##### this loops through a sequence e.g (strings, list, dictionary, tuple, lines in file, e.t.c)

```python
name = "Divine"
for a in name:
    print(a)

#here the program loops through the variable and print each letter in a different line
```

```python
for i in range(10):
    print(i)

# This prints the numbers from 0 to 9 because it excludes the last number
```

```python
for e in range(1, 21):
    print(e)

# Here, the program prints numbers from 1 to 20 because it excludes the last number in the range
```

```python
for k in range(5, 101, 5):
    print(k)

# This program prints the numbers from 5 to 100 with the interval as 5
```

```python
scores = [12, 47, 90, 87, 53, 28]
for d in scores:
    print(f"You scored {d} in the exams, but with the addition bonus mark it becomes {d + 5}")

#Here we implemented a list and then added 5 marks for all the students
```
