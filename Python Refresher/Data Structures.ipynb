# Data Structures in Python

##### Data structures is basically  a way data (values) are being stored, organised, and managed efficiently for seamless access and modification
##### They are 4 built in data structures in python, but users can specify structures explicitly such as stack, queues and linked list. Python built in data types are:
- List
- Tuple
- Set
- Dictionary

- List
##### A list is a collection of values which are ordered, mutable, and index-based. List are being declared using square bracket "[]"

```python
names = ["Divine", "Flourish", "Praise", "Victory", "Blessing"]

 # this slice the list first 3 items 
names[: 3]
```

```python
# adds an item at the end of a list
names.append("Joel")
names
```

```python
# removes an item from a list
names.remove("Blessing")
names
```

```python
# adds an item to the list at a position
names.insert(1, "Blessing")
names
```

```python
#reverse a list from end to beginning
names.reverse()
names
```

```python
# sort a list in ascending order
names.sort()
names
```

```python
# extend merges two list together
ages = [13, 25, 7, 2, 1, 11]
names.extend(ages)
names
```

```python
# copy makes a duplicate of the list
demographics = names.copy()
demographics
```

```python
# pop removes the last element in a list
# clear removes the entire items from a list
# count - counts the entire item in a list
```

- Tuple
##### Tuple is an immutable data structure meaning it cannot be modified after declaration and it is being declared by parenthesis "()". The only available method in tuple are:
- .count - counts a particular value in a tuple e.g names.count("Divine")
- .index - finds the position of an element in a tuple e.g names.index("Divine")
- .len - finds the total number of items in a tuple
- .max - finds the highest
- .sum - sums up all integers in a tuple
- .sort -sort a tuple 

```python
countries = ("Nigeria", "Ghana", "South Africa")
countries
```

- Dictionary
##### This is a very important data structure in python. It works with a key-value method just like an actual english dictionary where you search for a meaning of a word by actually navigating to the word. It is declared using curly braces "{}" 

```python
car_details = {
    "type": "SUV",
    "make": "Benz",
    "model": "C class",
    "color": "Blue"
}

car_details["type"]
```

```python
# dictionary with other dictionaries embedded into it

student_record = {
    "student_1":{
        "id": 101,
        "name": "Divine",
        "class": "SSS3",
        "is_brillian": True
    },
    "student_2":{
        "id": 102,
        "name": "Blessing",
        "class": "SSS1",
        "is_brillian": True  
    },
    "student_3":{
        "id": 103,
        "name": "Victory",
        "class": "JSS1",
        "is_brillian": True
    }
}

student_record["student_1"]
```
