# Error Handling in Python

##### Error handling is an effective way of preventing python code from crashing when the is any issue with the program. This gives the user a clear info about the issue and what to do next. Error handling is called **Exception** in python and it utilizes three major things (*try*, *except*, and *finally* which is optional)

```python
# A simple division operation
user_input_1 = int(input("Enter a number to divide: "))
divisor = int(input("Enter a divisor: "))

result = user_input_1 / divisor
```

##### From the above program, when we try to divide 20 by 0, we get an error but the message is vague. We can give the user more context by using exception

```python
try:
    user_input_1 = int(input("Enter a number to divide: "))
    divisor = int(input("Enter a divisor: "))

    result = user_input_1 / divisor
    print(f"The number {user_input_1} divided by {divisor} is equal to {result}")
except ZeroDivisionError:
    print(f"Error!, you cannot divide {user_input_1} by zero")
except ValueError:
    print(f"Error!, you can only perform divison on numbers")
finally:
    print("Calculation done")
```

##### Note that the **finally** block will run whether or not the condition is met
