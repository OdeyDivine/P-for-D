# Functions in Python

##### Functions are block of reusable code e.g the function **Print** can be used to print and output the terminal
##### functions uses the key word **def** followed by function name, parenthesis **()** and colon **:**

```python
# A simple function to print hello world
def hello():
    print("Hello world")

hello()
```

###### The above function doesn't take any parameter 

```python
# A function to add 2 numbers
def add(num1, num2):
    sum_num = num1 + num2
    return sum_num

add(5, 10)
```

###### The above function adds 2 numbers where are taken as a parameter

```python
# A function that divide 2 number
def divide(num1, num2):
    sum_num = num1 / num2
    return sum_num

divide(18, 3)
```

```python
# A function that multiply 2 number
def multiple(num1, num2):
    sum_num = num1 * num2
    return sum_num

multiple(5,3)
```

```python
# A function that subtracts from 2 numbers
def sub(num1, num2):
    sub_total = num1 - num2
    return sub_total

sub(40, 17)
```

```python
# A function that returns the square of the input
def power(num):
    return num**2

power(5)
```

```python
# A calculator function
def calculator():
    user_cal = int(input("Enter between 1 and 4 to calculate: "))
    option1 = int(input("Enter your first number: "))
    option2 = int(input("Enter your second number: "))
    if user_cal == 1:
        print(f"The total after adding is {add(option1, option2)}")
    elif user_cal == 2:
        print(f"The total after subtracting is {sub(option1, option2)}")
    elif user_cal == 3:
        print(f"The total after multiplying is {multiple(option1, option2)}")
    elif user_cal == 4:
        print(f"The total after dividing is {divide(option1, option2)}")
    else:
        print("Please enter a number between 1 and 4, our calculator is still simple")

calculator()
```

###### Remember we said functions are block of reusable code, in a calculator function we were able to call the add, sub, multiple, and divide function 

### Lambda 

##### Lambda enables us to write functions on a single line of code

```python
power_2 = lambda x: x**2
power_2(5)
```

### Map Function

```python
test_scores = [12,29,5,15,27,16]
bonus = lambda x: x + 5

upgraded_score = list(map(bonus, test_scores))
upgraded_score
```

###### The map function in the above cell is used to add and addition 5 marks to each student score which was saved in a list

### Filter Function

```python
exams_score = [30,29,18,56,90,17]
even_scores = list(filter(lambda x: x % 2 == 0, exams_score))
even_scores
```
