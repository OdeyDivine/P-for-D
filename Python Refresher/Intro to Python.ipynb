##### Floor Division of Floats and Integers
###### Floor division divides a number and truncate the decimal number (returns the number without the remainder after division)

```python
7 // 2
# this returns 3 because 2 divides 7 by 3 times with a reminder of 1
```

```python
2 // 7
# here we see the output as 0 because 7 cannot divide 2 at all
```

```python
-9 // 2
# for negative floor division, the remainder is being approximated
```

```python
10 // 2
```

```python
15 // -1.5
```

##### Modulus operation in Floats and Integers
###### modulus works alot like floor division, however modulus returns the remainder after division

```python
7 % 2
```

```python
7 % 2.5
```
