Pandas is made up of two main components which are Series, and DataFrame. A pandas serie is like an excel spreadsheet table, while a DataFrame is like the entire spreedsheet

```python
countries = pd.Series(["Nigeria", "Ghana", "Kenya", "Ivory Coast", "Niger", "Benin"])
countries
```

The above line created a pandas series which is like a single column in a spreadsheet

```python
columns_header = ["delivery_id", "item", "time_taken", "status"]
delivery_data = pd.DataFrame(
    [
        [101, "Rice", 35, "Delivered" ],
        [102, "Beans", 28, "Delivered"],
        [103, "Garri", 70, "Not Delivered"],
        [104, "Potato", 59, "Delivered"]
    ],
    columns = columns_header
)

delivery_data
```

The above code cell creates a Pandas dataframe using list but we have to create the columns name seperately which make the entire process really complicated

```python
dataset = pd.DataFrame(
    [
        {
            "name": "Odey Divine",
            "role": "Data Scientist",
            "salary": "$50,000"
        },
        {
            "name": "Blessing Odey",
            "role": "Product Designer",
            "salary": "$10,000"
        },
        {
            "name": "Victory Osang",
            "role": "Product Manager",
            "salary": "$7,000"
        }
    ]
)

dataset
```

The above code cell creates a DataFrame which acts as an spreedsheet with rows and columns. But here the data is stored in form of **JSON** or **Python Dictionaries** and this is a more compact way of data entry compared to List of List

```python
student_record = pd.DataFrame(
    {
        "student_id": [101, 102, 103, 104, 105, 106, 107, 108, 109],
        "name": ["Divine Odey", "Blessing Odey", "Victory Osang", "Daniel Bullem", "Maxwell Ogon", "Praise Bullem", " " , "Flourish Osang", "Joel Osang"],
        "score": [99, 97, 90, 89, 99, 96, 91, 91, 90]
    }
)

student_record
```

This is the most efficient way of creating a dataframe in pandas

```python
del student_record["score"]

# the above line of code deletes a column from a dataframe in pandas
student_record
```

```python
student_record.loc[0:5,"name"]
```

The above code cell selects from row from index 0 to index 5 inclusive, and only the name column. The .loc acts a location and remember when we said NumPy is being declared as rows before columns

```python
delivery_data[["delivery_id", "item" ]]
#this selects only the specified columns
```

```python
delivery_data.iloc[:, 0:2]
```

The above code cell uses *iloc* which I guess stands for index location, iloc is last value exclusive and it uses index for the columns instead of name of the column
