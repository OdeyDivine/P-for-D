```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([[4], [5], [6]])
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def plot_vectors(vectors, colors):
    """Plots 3D vectors from the origin using matplotlib."""
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111, projection='3d')

    # Set origin
    origin = np.zeros((3, 1))

    for i, vec in enumerate(vectors):
        # Ensure the vector is flattened to 1D with 3 elements
        v = np.array(vec).flatten()
        if len(v) < 3:
            # Pad with zeros if less than 3D
            v = np.pad(v, (0, 3 - len(v)), 'constant')
        elif len(v) > 3:
            v = v[:3]

        # Plot vector using quiver
        ax.quiver(0, 0, 0, v[0], v[1], v[2], color=colors[i % len(colors)], arrow_length_ratio=0.1, label=f'Vector {i+1}: {v}')

    # Adjust limits dynamically based on vectors
    all_vals = np.concatenate([np.array(v).flatten() for v in vectors])
    max_val = max(np.max(np.abs(all_vals)), 1)

    ax.set_xlim([-max_val, max_val])
    ax.set_ylim([-max_val, max_val])
    ax.set_zlim([-max_val, max_val])

    ax.set_xlabel('X axis')
    ax.set_ylabel('Y axis')
    ax.set_zlabel('Z axis')
    ax.set_title('3D Vector Visualization')
    ax.legend()
    ax.grid(True)
    plt.show()
```

```python
vectors = [a, b]
colors = ["r", "g"]
plot_vectors(vectors, colors)
```

```python
def vector_vector_multiplication(a, b):
  assert a.shape[0] == b.shape[0]
  result = 0.0
  n = a.shape[0]

  for i in range(n):
    result = result + a[i] * b[i]
  return result

vector_vector_multiplication(a, b)
```

```python
c = np.array([2,3,6,7])
d = np.array([
    [5, 6, 7, 9],
     [2, 7, 8, 5],
      [9, 4,2,6]
    ])

print(c.shape[0])
print(d.shape[1])
```

```python
def vector_matrix_multiplication(c, d):
  assert c.shape[0] == d.shape[1]
  n = d.shape[0]
  result = np.zeros(n)

  for i in range(n):
    result[i] = vector_vector_multiplication(c, d[i])
  return result

vector_matrix_multiplication(c, d)
```

```python
d.dot(c)
```
