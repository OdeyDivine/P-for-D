### Generating Random Numbers

```python
np.random.rand(3, 7)
```

The above code cell create float an array of floats between 0 and 1. Note that the value changes anytime we run the code, but we set it fixed using **seed**

```python
np.random.seed(123)
np.random.rand(4, 7)
```

```python
np.random.randn(3, 9)
```

This creates a normal distribution and it follow the rows and columns format




```python
np.random.randint(low = 10, high = 1001, size = (5, 9))
```

The above code creates and matrix starting from 10 and ending at 1000 and we also declared the shape of the array

```python
r = np.random.randint(2, 5, (2, 5))
m = np.random.randint(4, 9, (5, 2))

print(r)
print(m)
```

Here we have two matrix of shape (2 X 5) and shape(5 X 2) respectively. We can multiply them because their rows and columns opposite are equal. We multiply them below

```python
np.matmul(r, m)
```
