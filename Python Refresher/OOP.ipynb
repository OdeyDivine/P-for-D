# Object Oriented Programming in Python

##### Object-oriented programming (OOP) is a programming paradigm that organizes code around objects, combining data and behavior in a way that mirrors real-world entities. An object is a self-contained component with properties (data attributes) and behavior (methods).
##### A class is a user-defined prototype that defines a set of attributes and methods that characterise any class object

```python
class Person:
    pass

Person
```

##### In the code above, we created an empty class

```python
class Car:
    def __init__(self, model, color, brand):
        self.model = model
        self.color = color
        self.brand = brand

    def suv(self):
        print(f"The name of my car is {self.brand} and the model is {self.model}, and also colour {self.color}")

    def jeep(self, price):
        print(f"The name of my jeep is {self.model} and it is an {self.model}, and it cost ${price}")

Car("Camry", "Black", "Toyota").suv()
Car("Gwagon", "White", "Mercedes Benz").jeep(50000)

```

##### In the above cell, we declared a class called **Car** and we gave it a *model*, *color*, and *brand* by default. and then we created functions for **suv** and **jeep**

```python
class Details:
    nationality = "Nigerian"

    def __init__(self, region, age, state, name):
        self.region = region
        self.age = age
        self.state = state
        self.name = name

    def describe_person(self):
        print(f"My name is {self.name} and I am from {self.state} {Details.nationality}. I am {self.age} years old")

Details("South", 27, "Taraba", "Divine").describe_person()
```


## Inheritance In Python Class
##### Inheritance enables python define a class and inherit from another class

```python
class Animal:
    def __init__(self, sound, color):
        self.sound = sound
        self.color = color

    def describe(self):
        return f"The animal is colour {self.color}, and it {self.sound}"

result = Animal("bleats", 'black')
result.describe()
```

##### From the above code cell, we have been able to create a class called **Animal** in which we can inherit from

```python
class Dog(Animal):
    def __init__(self, sound, color, breed):
        # We can inherit all the properties from the Animal class and just enter breed because the Animal class doesn't have breed
        Animal.__init__(self, sound, color)
        self.breed = breed

    def info(self):
        return f"This is a {self.color} {self.breed}, and it {self.sound}"

result = Dog("barks", "White", "german-shepherd")
result.info()
    
```

##### We successfully inherited the **Animal** class for our newly defined class **Dog**. Note that we passed the properties from the parent class using  *Animal.\__init__\(self, sound, color)*  but this method isn't efficient (dynamic) so we will be using *super()* instead

```python
#Here we will create a class that inherit from the Dog class using super()
class Goat(Dog):
    def __init__(self, sound, color, breed, size):
        super().__init__(sound, color, breed)
        self.size = size

    def about(self):
        return f"This is a {self.color} {self.breed} which weight is {self.size}kg, and it {self.sound}"

result = Goat("bleats", "brown", 'he-goat', 29)
result.about()
```

```python
# Now we will create our last class for sheep and we won't be passing any extra property so we will using pass

class Sheep(Goat):
    pass

result = Sheep("bleats", "white", "mountain-sheep", 36)
result.about()
```


## Polymorphism In Python Class
##### This enables us create methods, operators, or functions with the same name but the perform different actions in a class or object. Take it as individual classes sharing the same name but have different parameters and we can also use this for inherited classes

```python
#Different classes with same method

class Dog:
    def __init__(self, breed):
        self.breed = breed

    def action(self):
        return "Barks"

class Lion:
    def __init__(self, breed):
        self.breed = breed

    def action(self):
        return "Roars"

class Goat:
    def __init__(self, breed):
        self.breed = breed

    def action(self):
        return "Bleats"

print(Goat("he-goat").action())
print(Lion("mountain-lion").action())
print(Dog('bull-dog').action())
```

##### From the above example we can see that both Dog, Goat, and Lion shares the same method called **action** but the results are different. Now lets do an example for an inherited class

```python
class Computer:
    def info(self):
        print("Window")

class Laptop(Computer):
    pass

class Phone(Computer):
    def info(self):
        print("Andriod")

class Ipad(Computer):
    def info(self):
        print("IOS")

com = Computer()
lappy = Laptop()
phone = Phone()
ipad = Ipad()
result = [com, lappy, phone, ipad]

for i in result:
    i.info()
```

##### From the above code cell we can see a clear example of polymorphism in action
