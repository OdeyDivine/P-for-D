# Object Oriented Programming in Python

##### Object-oriented programming (OOP) is a programming paradigm that organizes code around objects, combining data and behavior in a way that mirrors real-world entities. An object is a self-contained component with properties (data attributes) and behavior (methods).
##### A class is a user-defined prototype that defines a set of attributes and methods that characterise any class object

```python
class Person:
    pass

Person
```

##### In the code above, we created an empty class

```python
class Car:
    def __init__(self, model, color, brand):
        self.model = model
        self.color = color
        self.brand = brand

    def suv(self):
        print(f"The name of my car is {self.brand} and the model is {self.model}, and also colour {self.color}")

    def jeep(self, price):
        print(f"The name of my jeep is {self.model} and it is an {self.model}, and it cost ${price}")

Car("Camry", "Black", "Toyota").suv()
Car("Gwagon", "White", "Mercedes Benz").jeep(50000)

```

##### In the above cell, we declared a class called **Car** and we gave it a *model*, *color*, and *brand* by default. and then we created functions for **suv** and **jeep**

```python
class Details:
    nationality = "Nigerian"

    def __init__(self, region, age, state, name):
        self.region = region
        self.age = age
        self.state = state
        self.name = name

    def describe_person(self):
        print(f"My name is {self.name} and I am from {self.state} {Details.nationality}. I am {self.age} years old")

Details("South", 27, "Taraba)
```
